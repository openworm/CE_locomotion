Set the simulation parameters.

```python
args = dict(
    simduration=10,
    simtransient=10,
    duration=10,
    transient=10,
    maxGens=10,
    popSize=26,
    RandSeed=81235,
    modelName="W2DCE",
    modelFolder="Worm2D",
    outputFolderName=outputFolderName,
    doEvol=True,
    overwrite=True,
    checkPointInterval=5,
    reRand=True,
    doPlotEvol=True,
    doNML=False,
    doCPT=True,
    evoType="EvoCE",
    doReverse=2,
    SRType="None",
    ABLevel=3,
    doForwardFirst=True,
    # SRType="SR_TRANS_CONTRACT",
    # SRType = "SR_TRANS_STRETCH",
    # SRType = "SR_TRANS_ABS",
    # SRType = "SR_TRANS_NEG"
    randInitState = True,
    doOrigMuscInput = False,
    doOrigSRInput = False,
    doLegacy = False,
    showPlot = True
    # MutVar = 0.5
)
```

Print part of json out for inspection:

```python
import json
with open(outputFolderName + '/worm_data_evo.json') as json_file:
    json_data = json.load(json_file)
print(json_data['Nervous system']['Chemical weights']['value'])
```

Modify and value in the json:

```python
with open(outputFolderName + '/worm_data_evo.json') as json_file:
    json_data = json.load(json_file)
print(json_data['Nervous system']['Chemical weights']['value'][1])
json_data['Nervous system']['Chemical weights']['value'][1]['weight'] = 0
print(json_data['Nervous system']['Chemical weights']['value'][1])
```

Back up the original json and run simulation of the new json:

```python
import shutil
shutil.copyfile(outputFolderName + '/worm_data_evo.json', 
                outputFolderName + '/worm_data_evo_orig.json')
with open(outputFolderName + '/worm_data_evo.json', "w") as f:
    json.dump(json_data, f)
args['doEvol'] = False
args['modelName'] = "W2DSR"
args['simduration'] = 20
run(**args)
```

Change weight back to its original value

```python
json_data['Nervous system']['Chemical weights']['value'][1]['weight'] = -15.071462336914168
print(json_data['Nervous system']['Chemical weights']['value'][1])
```

Save the JSON and run the simulation again to produce the original activity:

```python
with open(outputFolderName + '/worm_data_evo.json', "w") as f:
    json.dump(json_data, f)
args['doEvol'] = False
args['modelName'] = "W2DSR"
args['simduration'] = 20
run(**args)
```
