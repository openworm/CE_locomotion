After a simulation has been performed the output folder contains various files. These files are updated every time the code is run and should be reloaded in this notebook after every simulation to view the most recent version. The fitness.dat file shows the phenotype fitness history for the generations so far. The first column is the generation number, the second the fitness of the best phenotype, the third the average fitness of the population and the fourth the standard deviation of the population fitness. This file updates dynamically and can be reloaded during an evolution to examine the progress. The evolution can be interrupted if desired and a simulation performed or the evolution resumed from the last checkpoint (by default every five generations):

```python
with open(outputFolderName + '/fitness.dat') as f:
    print(f.read())
```

When the genetic algorithm optimization is performed, some model parameters are fined tuned to produce a particular behaviour for the phenotype. The following figure (top left panel) shows how the fitness, (between zero and unity) in log scale, for the best phenotype in the population and the population average phenotype varies over successive generations. The other five panels show how the values of the evolvable model parameters vary over successive generations. The optimizable parameters are shown in the bottom right panel. For example there may be parameters associated with chemical connection weights in the nervous system (denoted ns_chemcons), parameters associated with the stretch receptor gains (sr_head_gain, sr_vnc_gain) and others associated with the VNC nervous system to muscle connectivity structure (denoted vnc_18_dd_vda_vdp, for example). Many parameters will have the same name, for example there may be multiple "ns_cells_bias", these are bias parameters for different cells, or different sets of cells, since the same evolvable parameter may be used across multiple cells. The figure shows their actual values (top right), percent change from the intial state (middle left), signed log of the percent change (bottom left) all averaged across the whole population. The percent change values for the best phenotype is also shown (middle right).
